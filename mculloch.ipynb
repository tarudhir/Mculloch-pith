```python
import numpy as np
```

```python
class McCullochPittsNeuron:
    def __init__(self, weights, threshold):
        self.weights = np.array(weights)
        self.threshold = threshold

    def activate(self, inputs):
        net_input = np.dot(self.weights, inputs)
        return 1 if net_input >= self.threshold else 0
```

```python
inputs = [(0, 0), (0, 1), (1, 0), (1, 1)]

def test_gate(gate_name, neuron, inputs):
    print(f"\n{gate_name} GATE")
    print("A  B  | Output")
    print("----------------")
    for x in inputs:
        output = neuron.activate(x)
        print(f"{x[0]}  {x[1]}  |   {output}")
```

```python
or_gate = McCullochPittsNeuron([1, 1], 1)
test_gate("OR", or_gate, inputs)
```

```python
not_gate = McCullochPittsNeuron([-1], 0)

print("NOT GATE")
print("A | Output")
print("----------")

for x in [(0,), (1,)]:
    print(f"{x[0]} |   {not_gate.activate(x)}")
```

```python
nand_gate = McCullochPittsNeuron([-1, -1], -1)
test_gate("NAND", nand_gate, inputs)
```

```python

nor_gate = McCullochPittsNeuron([-1, -1], 0)
test_gate("NOR", nor_gate, inputs)
```

```python
def xor_gate(a, b):
    or_neuron = McCullochPittsNeuron([1, 1], 1)
    nand_neuron = McCullochPittsNeuron([-1, -1], -1)
    and_neuron = McCullochPittsNeuron([1, 1], 2)

    or_output = or_neuron.activate([a, b])
    nand_output = nand_neuron.activate([a, b])

    return and_neuron.activate([or_output, nand_output])

print("XOR GATE")
print("A  B  | Output")
print("----------------")

for a, b in inputs:
    print(f"{a}  {b}  |   {xor_gate(a, b)}")
```

```python
def xnor_gate(a, b):
    return 1 - xor_gate(a, b)

print("XNOR GATE")
print("A  B  | Output")
print("----------------")

for a, b in inputs:
    print(f"{a}  {b}  |   {xnor_gate(a, b)}")
```
